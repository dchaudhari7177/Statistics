```python
import os, os.path
import cv2
import numpy as np
path="D:\\backup\CHARACTERS"
for root, dirs, files in os.walk(path):
    for f in files:
        fullpath = os.path.join(root, f)
        if os.path.splitext(fullpath)[1] == '.jpg':
            base=os.path.basename(f)
            os.path.splitext(base)
            FName=os.path.splitext(base)[0]
            img = cv2.imread(fullpath)
            img=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
            img = cv2.bitwise_not(img)
            th2 = cv2.adaptiveThreshold(img,255, cv2.ADAPTIVE_THRESH_MEAN_C,cv2.THRESH_BINARY,15,-2)
           #ret,th2 = cv2.threshold(img,127,255,cv2.THRESH_BINARY_INV)
            #cv2.imshow("th2", th2)
            cv2.imwrite("th2.jpg", th2) 
            horizontal = th2
            vertical = th2
            rows,cols = horizontal.shape
            horizontalsize = cols // 25
            horizontalStructure = cv2.getStructuringElement(cv2.MORPH_RECT, (horizontalsize,1))
            horizontal = cv2.erode(horizontal, horizontalStructure, (-1, -1))
            horizontal = cv2.dilate(horizontal, horizontalStructure, (-1, -1))
#cv2.imshow("horizontal", horizontal)
            cv2.imwrite("horizontal.jpg", horizontal)
            cv2.waitKey(10)
            cv2.destroyAllWindows()
            verticalsize = rows // 25
            verticalStructure = cv2.getStructuringElement(cv2.MORPH_RECT, (1, verticalsize))
            vertical = cv2.erode(vertical, verticalStructure, (-1, -1))
            vertical = cv2.dilate(vertical, verticalStructure, (-1, -1))
#cv2.imshow("vertical", vertical)
            cv2.imwrite("vertical.jpg", vertical)
            img3=cv2.add(horizontal,vertical)
#cv2.imshow("emptygrid",img3)
            cv2.imwrite("emptygrid.jpg",img3)
            kernel = np.ones((2, 2), dtype = "uint8")
            img5=cv2.dilate(img3,kernel,iterations=2)
#cv2.imshow("image",img5)
            cv2.imwrite("image.jpg",img5)
            img4=cv2.subtract(th2,img5)
            kernel = np.ones((2,2), dtype = "uint8")
            erosion = cv2.erode(img4,kernel,iterations = 1)
#cv2.imshow("erosion",erosion)
            cv2.imwrite("erosion.jpg",erosion)
            dilation = cv2.dilate(erosion ,kernel,iterations =1)
#Scv2.imshow("dilation",dilation)
            cv2.imwrite("dilation.jpg",dilation)
            clear = cv2.bitwise_not(dilation)
#cv2.imshow("clear",clear)
            (x,image1)=cv2.threshold(clear,190,255,cv2.THRESH_BINARY_INV)
            image1=cv2.medianBlur(image1,5)
            image1=cv2.morphologyEx(image1, cv2.MORPH_OPEN, cv2.getStructuringElement(cv2.MORPH_RECT,(3,3)))
            os.chdir("D:\\backup\CHARACTERS\linesremoved")
            f1=os.path.join(f,".jpg")
            cv2.imwrite(f1,image1)

           
```
