```python
Consider a die is thrown 600 times with following results.

observed = [115, 47, 41, 101, 210, 96]

Is the die unbiased at 10% significance level?
```

```python
import numpy as np
from scipy.stats import chi2
observed = np.array([115, 47, 41, 201, 110, 96])# Observed frequencies
expected = np.array([100, 100, 100, 100, 100, 100])# Expected frequencies (assuming a fair die)
chi2_stat = np.sum((observed - expected)**2 / expected)# Calculate chi-square statistic
df = len(observed) - 1  # Degrees of freedom (number of categories - 1)
# Critical value for 10% significance level
critical_value = chi2.ppf(0.90, df)
p_value = 1 - chi2.cdf(chi2_stat, df)# p-value
print(f"Chi-squared Statistic: {chi2_stat}")
print(f"Critical Value at 10% significance level: {critical_value}")
print(f"p-value: {p_value}")
if chi2_stat < critical_value:
    print("Fail to reject the null hypothesis: The die is unbiased.")
else:
    print("Reject the null hypothesis: The die is biased.")
```

```python
question: specification. A pin may meet the specification, or it may be too thin or too thick. 
Pins are sampled from each machine, and the number of pins in each category is counted.
Table below presents the results. Use the data in Table to test the null hypothesis that the proportions of pins that are
too thin, OK, or too thick are the same for all the machines.
Table

data = np.array([
    [10, 102, 8],   # Machine 1
    [34, 161, 5],   # Machine 2
    [12, 79, 9],    # Machine 3
    [10, 60, 10]    # Machine 4
])
```

```python
import numpy as np
import pandas as pd
from scipy.stats import chi2
from scipy.stats import chi2_contingency
# Define the observed data
data = np.array([
    [10, 102, 8],   # Machine 1
    [34, 161, 5],   # Machine 2
    [12, 79, 9],    # Machine 3
    [10, 60, 10]    # Machine 4
])
# Create a DataFrame for better visualization (optional)
df = pd.DataFrame(data, columns=['Too Thin', 'OK', 'Too Thick'], 
                  index=['Machine 1', 'Machine 2', 'Machine 3', 'Machine 4'])
print("Observed Data:\n", df)
chi2_stat, p_value, dof, expected = chi2_contingency(data)# Perform the Chi-Square test
print("\nChi-Square Statistic:", chi2_stat)
print("P-Value:", p_value)
print("Degrees of Freedom:", dof)
print("Expected Frequencies:\n", expected)
alpha = 0.05
if chi2_stat > chi2.ppf(1 - alpha, dof):
    print("Reject the null hypothesis: There is a significant difference.")
else:
    print("Fail to reject the null hypothesis: No significant difference.")
```

```python
Chi-Square Test and Test of Independence, Analyzing the relationship between gender and preference for a product purchase
```

```python
import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency
import matplotlib.pyplot as plt

# Create a contingency table
data = np.array([[30, 10],  # Male
                 [20, 30]]) # Female

# Display the contingency table as a DataFrame for clarity
contingency_table = pd.DataFrame(data, 
                                  columns=['Purchased', 'Not Purchased'], 
                                  index=['Male', 'Female'])
print("Contingency Table:\n", contingency_table)

# Perform the Chi-Square test
chi2_stat, p_value, dof, expected = chi2_contingency(data)

# Display results
print("\nChi-Square Statistic:", chi2_stat)
print("P-Value:", p_value)
print("Degrees of Freedom:", dof)
print("Expected Frequencies:\n", expected)

# Determine significance level
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis: There is a significant association between gender and product preference.")
else:
    print("Fail to reject the null hypothesis: No significant association between gender and product preference.")
```

```python
using Chi-Square Test, Analyzing the association between vaccination and recovery of the patients considering COVID data.
```

```python
import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency
import matplotlib.pyplot as plt
# Create a contingency table
data = np.array([[10, 30],    # Vaccinated
                 [10, 40]])   # Not Vaccinated
# Display the contingency table as a DataFrame for clarity
contingency_table = pd.DataFrame(data, 
                                  columns=['Recovered', 'Not Recovered'], 
                                  index=['Vaccinated', 'Not Vaccinated'])
print("Contingency Table:\n", contingency_table)
# Perform the Chi-Square test
chi2_stat, p_value, dof, expected = chi2_contingency(data)
# Display results
print("\nChi-Square Statistic:", chi2_stat)
print("P-Value:", p_value)
print("Degrees of Freedom:", dof)
print("Expected Frequencies:\n", expected)
# Determine significance level
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis: There is a significant association between vaccination and recovery.")
else:
    print("Fail to reject the null hypothesis: No significant association between vaccination and recovery.")

```
