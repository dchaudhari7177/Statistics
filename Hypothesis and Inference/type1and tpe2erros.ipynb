```python
A test is made of the hypotheses H0 :μ ≤ 10 versus
H1 :μ > 10. For each of the following situations, determine
whether the decision was correct, a type I error
occurred, or a type II error occurred.
a. μ = 8, H0 is rejected.
b. μ = 10, H0 is not rejected.
c. μ = 14, H0 is not rejected.
d. μ = 12, H0 is rejected
```

```python
# Define the function to check the decision against the true mean
def evaluate_decision(true_mean, decision, rejection_threshold=10):
    if decision == "reject":
        if true_mean <= rejection_threshold:
            return "Type I Error (H0 rejected when it is true)"
        else:
            return "Correct Decision (H1 is true)"
    else:  # decision == "not reject"
        if true_mean > rejection_threshold:
            return "Type II Error (H0 not rejected when it is false)"
        else:
            return "Correct Decision (H0 is true)"

# Define the situations
situations = [
    {"true_mean": 8, "decision": "reject"},   # Situation a
    {"true_mean": 10, "decision": "not reject"}, # Situation b
    {"true_mean": 14, "decision": "not reject"}, # Situation c
    {"true_mean": 12, "decision": "reject"},  # Situation d
]

# Evaluate each situation
for i, situation in enumerate(situations, start=1):
    true_mean = situation["true_mean"]
    decision = situation["decision"]
    result = evaluate_decision(true_mean, decision)
    print(f"Situation {chr(96 + i)}: True Mean = {true_mean}, Decision = {decision} -> {result}")
```
