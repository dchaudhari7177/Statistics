```python
import pandas as pd

titanic_data = pd.read_csv('Titanic-Dataset.csv')

central_tendency = {
    'Feature': ['PassengerId', 'Survived', 'Pclass', 'Name', 'Sex', 'Age', 'SibSp', 'Parch', 'Ticket', 'Fare', 'Cabin', 'Embarked'],
    'Measure of Central Tendency': ['Mode', 'Mode', 'Mode', 'Mode', 'Mode', 'Mean', 'Mean', 'Mean', 'Mode', 'Mean', 'Mode', 'Mode'],
    'Value': [
        titanic_data['PassengerId'].mode()[0], 
        titanic_data['Survived'].mode()[0], 
        titanic_data['Pclass'].mode()[0], 
        titanic_data['Name'].mode()[0], 
        titanic_data['Sex'].mode()[0], 
        titanic_data['Age'].mean(), 
        titanic_data['SibSp'].mean(), 
        titanic_data['Parch'].mean(), 
        titanic_data['Ticket'].mode()[0], 
        titanic_data['Fare'].mean(), 
        titanic_data['Cabin'].mode()[0], 
        titanic_data['Embarked'].mode()[0]
    ]
}

central_tendency_df = pd.DataFrame(central_tendency)

numeric_columns = titanic_data.select_dtypes(include=['float64', 'int64'])

std_dev_range = {
    'Feature': numeric_columns.columns,
    'Standard Deviation': numeric_columns.std(),
    'Range': numeric_columns.max() - numeric_columns.min()
}

std_dev_range_df = pd.DataFrame(std_dev_range)


summary_stats = pd.concat([central_tendency_df, std_dev_range_df[['Standard Deviation', 'Range']]], axis=1)

summary_stats
```
