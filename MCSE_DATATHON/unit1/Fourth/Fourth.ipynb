```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset('titanic')

print(df.columns)

df = df.dropna(subset=['age', 'fare'])  

fig, axes = plt.subplots(2, 2, figsize=(14, 12))

sns.histplot(df['age'], bins=30, kde=True, ax=axes[0, 0], color='blue')
axes[0, 0].set_title('Histogram of Age')

sns.boxplot(x=df['age'], ax=axes[0, 1], color='blue')
axes[0, 1].set_title('Box Plot of Age')

sns.histplot(df['fare'], bins=30, kde=True, ax=axes[1, 0], color='green')
axes[1, 0].set_title('Histogram of Fare')

sns.boxplot(x=df['fare'], ax=axes[1, 1], color='green')
axes[1, 1].set_title('Box Plot of Fare')

plt.tight_layout()
plt.show()
```
