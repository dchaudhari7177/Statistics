```python
import pandas as pd

# Load the dataset
titanic_data = pd.read_csv("Titanic-Dataset.csv")

# Check for missing values
missing_values = titanic_data.isnull().sum()

# Handle missing values

# 1. For missing 'Age' values, we can fill with the median age (as age is numerical and doesn't follow a specific pattern)
titanic_data['Age'] = titanic_data['Age'].fillna(titanic_data['Age'].median())

# 2. For missing 'Cabin' values, we can drop the column since it has too many missing values or fill it with 'Unknown'
titanic_data['Cabin'] = titanic_data['Cabin'].fillna('Unknown')

# 3. For missing 'Embarked' values, we can fill them with the mode of the column (most frequent value)
titanic_data['Embarked'] = titanic_data['Embarked'].fillna(titanic_data['Embarked'].mode()[0])

# 4. For missing 'Fare' values, we can fill them with the median fare
titanic_data['Fare'] = titanic_data['Fare'].fillna(titanic_data['Fare'].median())

# 5. For missing 'SibSp' and 'Parch', we can fill with 0, assuming that the passenger was not traveling with family
titanic_data['SibSp'] = titanic_data['SibSp'].fillna(0)
titanic_data['Parch'] = titanic_data['Parch'].fillna(0)

# Handle inconsistent data formatting

# 1. Ensure the 'Sex' column is consistent (lowercase)
titanic_data['Sex'] = titanic_data['Sex'].str.lower()

# Check for duplicate rows and remove them if necessary
titanic_data = titanic_data.drop_duplicates()

# Convert columns to appropriate data types if needed (e.g., 'Survived' should be int)
titanic_data['Survived'] = titanic_data['Survived'].astype(int)

# Verify data types of all columns after cleaning
data_types = titanic_data.dtypes

# Check missing values again after cleaning
missing_values_after_cleaning = titanic_data.isnull().sum()

# Save the cleaned data to a new CSV file
titanic_data.to_csv("Cleaned_Titanic_Dataset.csv", index=False)

# Display the cleaned dataset's first few rows
print(titanic_data.head())
```
