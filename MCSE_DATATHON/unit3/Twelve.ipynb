```python
import pandas as pd
import numpy as np
import seaborn as sns

df = sns.load_dataset('titanic')

print("Columns in the Titanic dataset:", df.columns)

df['Family_Size'] = df['sibsp'] + df['parch']

df['Is_Alone'] = np.where(df['Family_Size'] == 0, 1, 0)

bins = [0, 12, 19, 59, np.inf]
labels = ['Child', 'Teen', 'Adult', 'Senior']
df['Age_Group'] = pd.cut(df['age'], bins=bins, labels=labels)

df['Fare_Pclass'] = df['fare'] / df['pclass']

if 'cabin' in df.columns:
    df['Cabin_Known'] = np.where(df['cabin'].isnull(), 0, 1)
else:
    df['Cabin_Known'] = np.nan 

fare_bins = [0, 50, 100, 150, 200, np.inf]
fare_labels = ['Low', 'Medium', 'High', 'Very High', 'Extremely High']
df['Fare_Binned'] = pd.cut(df['fare'], bins=fare_bins, labels=fare_labels)

print("\nNewly created features:\n", df[['Family_Size', 'Is_Alone', 'Age_Group', 'Fare_Pclass', 'Cabin_Known', 'Fare_Binned']].head())
```
